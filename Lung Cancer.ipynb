```python
X = df_new.drop('LUNG_CANCER', axis = 1)
y = df_new['LUNG_CANCER']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size= 0.25, random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model= DecisionTreeClassifier(criterion='entropy', random_state=0)
dt_model.fit(X_train, y_train)
```

```python
y_dt_prediction= dt_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report
dt_cr=classification_report(y_test, y_dt_prediction)
print(dt_cr)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_estimator(dt_model, X_test, y_test, cmap='Blues')
plt.title("Figure: Confusion Matrix for Decision Tree")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier()
rf_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report
y_rf_pred = rf_model.predict(X_test)
rf_cr=classification_report(y_test, y_rf_pred)
print(rf_cr)
```

```python
from xgboost import XGBClassifier
xgb_model = XGBClassifier()
xgb_model.fit(X_train, y_train)
```

```python
y_xgb_pred = xgb_model.predict(X_test)
xgb_cr=classification_report(y_test, y_xgb_pred)
print(xgb_cr)
```

```python
ConfusionMatrixDisplay.from_estimator(xgb_model, X_test, y_test, cmap='Blues')
plt.title("Figure: Confusion Matrix for XGBoost")
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import numpy as np
from sklearn.linear_model import LogisticRegression

y_true = y_test

model = LogisticRegression()

y_pred = model.fit(X_test, y_test).predict(X_test)


cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(10, 8))
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap=plt.cm.Blues, xticks_rotation='vertical')
plt.title('Confusion Matrix for Lung Cancer Prediction')
plt.grid(False)
plt.show()
```

```python
symptom_columns = [col for col in df.columns if col not in ['AGE', 'LUNG_CANCER']]

mean_values = df.groupby('LUNG_CANCER')[symptom_columns].mean().T

mean_values.columns = ['No Lung Cancer', 'Lung Cancer']

plt.figure(figsize=(12, 6))
sns.lineplot(data=mean_values, markers=True, dashes=False)
plt.title('Symptom Frequency by Lung Cancer Status')
plt.xlabel('Symptoms')
plt.ylabel('Proportion (Mean Value)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
